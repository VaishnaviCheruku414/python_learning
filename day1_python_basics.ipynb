# 3. 📦 Variables: labelled boxes

### 📖 Idea
A **variable** is a **labelled box** where the computer keeps something so you can use it later.

```
  name  ──►  [ "Priya" ]
  age   ──►  [   20    ]
```

The `=` sign does NOT mean "equal" here. It means **"put this into the box"**.

````python
name = "Priya"        # a box called name holds the text "Priya"
age = 20              # a box called age holds the number 20
height = 5.4          # a box called height holds a decimal number
is_student = True     # a box called is_student holds True

print(name)
print(age)
print(height)
print(is_student)
````

### 🔄 Boxes can change
You can put a new thing in the same box. The old thing is gone.

````python
score = 50
print(score)

score = 75        # new value replaces the old one
print(score)

score = score + 5  # take the current value, add 5, put it back
print(score)
````

### 📌 Rules for naming boxes

| Rule | ✅ Good | ❌ Bad |
|---|---|---|
| No spaces | `my_name` | `my name` |
| Cannot start with a number | `name1` | `1name` |
| Only letters, numbers and `_` | `total_marks` | `total-marks`, `total$` |
| Capital letters are different | `Age` and `age` are two different boxes | |
| Don't use Python's own words | `total` | `print`, `if`, `for` |

💡 **Style tip:** use small letters with underscores, like `student_name`. Choose names that explain themselves.

# 4. 🧬 Data types: what kind of thing is in the box?

### 📖 Idea
Every value has a **type**, like labels on food: vegetable, fruit, snack. The type decides what you can do with it.

| Type | Name in Python | What it holds | Example |
|---|---|---|---|
| Text | `str` (string) | words and letters | `"hello"` |
| Whole number | `int` (integer) | numbers without decimals | `20` |
| Decimal number | `float` | numbers with decimals | `5.4` |
| True / False | `bool` (boolean) | yes or no | `True`, `False` |

Use `type()` to ask Python "what is this?".

````python
print(type("hello"))
print(type(20))
print(type(5.4))
print(type(True))
````

### 🔁 Changing types (type conversion)
You can convert one type into another.

````python
print(int("25"))      # text to whole number  -> 25
print(float("3.5"))   # text to decimal       -> 3.5
print(str(100))       # number to text        -> "100"
print(int(7.9))       # decimal to whole      -> 7 (cuts off, does not round)
````

### ⚠️ Common mistake
`"5"` (text) and `5` (number) look the same but are **different things**.

````python
print("5" + "5")   # text + text joins them together -> 55
print(5 + 5)       # number + number adds them       -> 10
````

# 5. 🧮 Maths: the computer as a calculator

### 📖 Operators

| Symbol | Meaning | Example | Result |
|---|---|---|---|
| `+` | add | `10 + 3` | `13` |
| `-` | subtract | `10 - 3` | `7` |
| `*` | multiply | `10 * 3` | `30` |
| `/` | divide (always gives decimals) | `10 / 3` | `3.333...` |
| `//` | divide, keep only whole part | `10 // 3` | `3` |
| `%` | remainder (what is left over) | `10 % 3` | `1` |
| `**` | power | `10 ** 3` | `1000` |

````python
a = 10
b = 3

print("Add:", a + b)
print("Subtract:", a - b)
print("Multiply:", a * b)
print("Divide:", a / b)
print("Whole-number divide:", a // b)
print("Remainder:", a % b)
print("Power:", a ** b)
````

### 🧠 Order of operations (same as BODMAS in school)
Brackets first, then powers, then multiply/divide, then add/subtract.

````python
print(2 + 3 * 4)     # 14, because multiplication goes first
print((2 + 3) * 4)   # 20, because brackets go first
````

### 💡 Where is `%` useful?
- **Even or odd:** `number % 2` is `0` for even numbers, `1` for odd numbers.
- **Every 5th item, clock time, cycles** and many interview questions.

````python
print(10 % 2)   # 0 -> even
print(7 % 2)    # 1 -> odd
````

# 6. 🔤 Strings: working with text

### 📖 Idea
A **string** is text. Think of it as a **train of letters**, and each coach has a seat number.

**Python counts from 0, not 1!** It can also count backwards from the end using negative numbers.

```
 letter :   p   y   t   h   o   n
 forward:   0   1   2   3   4   5
 backward: -6  -5  -4  -3  -2  -1
```

````python
word = "python"

print(word[0])     # first letter  -> p
print(word[3])     # fourth letter -> h
print(word[-1])    # last letter   -> n
print(len(word))   # how many letters -> 6
````

### ✂️ Slicing (cutting a piece of text)
`word[start:stop]` takes letters from `start` up to, but **not including**, `stop`.

````python
print(word[0:3])   # pyt
print(word[2:])    # thon  (from position 2 to the end)
print(word[:4])    # pyth  (from the start to position 3)
print(word[::-1])  # nohtyp (reversed!)
````

### 🛠️ Useful string tools (called *methods*)

````python
text = "  Hello Data World  "

print(text.upper())          # ALL CAPITALS
print(text.lower())          # all small letters
print(text.strip())          # removes extra spaces at both ends
print(text.replace("World", "Python"))   # swap one word for another
print("Data" in text)        # is "Data" inside the text? -> True
print(text.split())          # cut into a list of words
print("python".capitalize()) # Python
print("python".title())      # Python (each word starts with a capital)
````

### 🔗 Joining and repeating

````python
print("I love " + "Python")   # + joins text together
print("Ha" * 3)               # * repeats text -> HaHaHa
````

### ⚠️ Remember
Strings **cannot be changed in place**. Methods like `.upper()` give you a **new** string. To keep it, put it in a variable:
`word = word.upper()`

# 7. 🎨 f-strings: mixing text and variables

### 📖 Idea
An **f-string** is a fill-in-the-blanks sentence. Put an `f` before the quotes, and write variable names inside `{ }`. Python fills in the values.

It is the cleanest way to build sentences, and you will use it **all the time**.

````python
name = "Priya"
age = 20
print(f"My name is {name} and I am {age} years old")
````

### 💡 You can even do maths and formatting inside `{ }`

````python
price = 49.5
quantity = 3

print(f"Total = {price * quantity}")
print(f"Price with 2 decimals = {price:.2f}")      # .2f means 2 digits after the decimal point
print(f"Big number = {1234567:,}")                 # adds commas -> 1,234,567
````

# 8. ⌨️ Input: asking the user questions

### 📖 Idea
`input()` **pauses** the program and waits for the user to type something. It is like the computer **asking you a question**.

### ⚠️ Very important
`input()` **always gives text (a string)**, even if you type a number. To do maths, convert it with `int()` or `float()`.

> 💡 When you run these cells, a small box appears. Type your answer and press **Enter**.

````python
user_name = input("What is your name? ")
print(f"Nice to meet you, {user_name}!")
````

````python
num = int(input("Enter a number: "))   # int() converts the text into a whole number
print(f"Double of {num} is {num * 2}")
````

### ⚠️ What if we forget `int()`?
Then Python treats your number as **text**, and `*` repeats it instead of multiplying!

````python
num_text = input("Enter a number: ")
print(num_text * 2)     # if you typed 5 you get 55, not 10
````

💡 For decimals use `float(input(...))`.

# 10. 🚨 Common errors (don't panic!)

### 📖 Idea
A **red error message** is not a failure. It is Python politely telling you what to fix. **Read the LAST line** of the error. It names the problem.

| Error name | What it means | Example | Fix |
|---|---|---|---|
| `SyntaxError` | You broke a writing rule | missing quote or bracket | check quotes `" "` and brackets `( )` |
| `NameError` | Python doesn't know that name | typo in a variable name | check spelling, make sure the variable was created first |
| `TypeError` | Mixing things that don't mix | `"5" + 5` | convert types with `int()` or `str()` |
| `ValueError` | Right type, wrong content | `int("hello")` | make sure the text really looks like a number |
| `ZeroDivisionError` | You divided by 0 | `10 / 0` | never divide by 0 |
| `IndentationError` | Wrong spaces at the start of a line | extra space before `print` | remove the extra spaces |

### 🧪 Break it on purpose
Remove the `#` from **one line at a time**, run the cell, read the error, then put the `#` back.

````python
# print("Hello)           # SyntaxError: missing closing quote
# print(namee)             # NameError: 'namee' is not defined
# print("5" + 5)           # TypeError: can only concatenate str to str
# print(int("hello"))      # ValueError: invalid literal for int()
# print(10 / 0)            # ZeroDivisionError: division by zero
````

### 💡 Debugging checklist
1. Read the **last line** of the error.
2. Look at the **line number** it points to.
3. Check **spelling, quotes, brackets, and spaces**.
4. Print your variables to see what is really inside them.
5. Still stuck? Copy the error message into a search engine or ask someone.

# 11. 📋 Cheat sheet (read this before interviews!)

| Topic | Remember this | Example |
|---|---|---|
| Print | shows output | `print("Hi")` |

<details>
<summary><b>1. What is a variable in Python?</b></summary>

A named place in memory that stores a value. We create it with `=`, like `age = 20`. Python decides the type automatically (this is called dynamic typing).
</details>

<details>
<summary><b>2. What are the basic data types in Python?</b></summary>

`int` (whole numbers), `float` (decimals), `str` (text), and `bool` (`True`/`False`).
</details>

<details>
<summary><b>3. What is the difference between <code>/</code> and <code>//</code>?</b></summary>

`/` gives the exact result with decimals (`7 / 2 = 3.5`). `//` gives only the whole-number part (`7 // 2 = 3`).
</details>

<details>
<summary><b>4. What does <code>%</code> do?</b></summary>

It gives the remainder after division. `7 % 2 = 1`. It is often used to check even or odd.
</details>

<details>
<summary><b>5. What is the difference between <code>5</code> and <code>"5"</code>?</b></summary>

`5` is an integer, and `"5"` is a string. `5 + 5 = 10`, but `"5" + "5" = "55"`.
</details>

<details>
<summary><b>6. What does <code>input()</code> return?</b></summary>

Always a string. To use it as a number, convert it with `int()` or `float()`.
</details>

<details>
<summary><b>7. How do you write a comment?</b></summary>

Start the line with `#`. Python ignores it.
</details>

<details>
<summary><b>8. How do you reverse a string?</b></summary>

Use slicing with a step of -1: `word[::-1]`.
</details>

<details>
<summary><b>9. What is an f-string?</b></summary>

A string that starts with `f` and lets us place variables or expressions inside `{ }`, like `f"Hello {name}"`.
</details>

<details>
<summary><b>10. Why does Python start counting from 0?</b></summary>

Indexes measure the **distance from the start**. The first item is 0 steps from the start. It is a common convention in most programming languages.
</details>
